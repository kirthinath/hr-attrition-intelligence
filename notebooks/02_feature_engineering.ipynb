# HR Attrition Intelligence - 02. Feature Engineering

This notebook documents the creation and verification of the 10 custom, domain-driven HR business features designed to improve model sensitivity and provide executive explainability.

### The 10 Custom Features:
1. **Income Band**: Bins salary into corporate tiers.
2. **Tenure Group**: Clusters employees by experience cohorts.
3. **Promotion Delay Risk**: Relative stagnation metric.
4. **Overtime Risk Score**: Combines overtime and engagement/work-life indicators.
5. **Work-Life Balance Risk**: Synthesizes travel burden and balance ratings.
6. **Satisfaction Composite Score**: Consolidated job environment happiness indicator.
7. **Manager Stability Score**: Ratio of tenure under current manager.
8. **Career Growth Risk Indicator**: Stagnation measure relative to working years.
9. **Travel Burden Score**: Combines distance and travel frequency.
10. **Burnout Risk Index**: Multi-factor stress index.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

## 1. Load Cleaned Dataset

```python
df = pd.read_csv("../data/processed/cleaned_hr_data.csv")
print(f"Cleaned dataset columns: {df.shape[1]}")
df.head()
```

## 2. Verify Engineered Feature Correlations

Let's check how the new features correlate with the target `attrition` column compared to baseline indicators.

```python
cols = [
    "attrition", "overtime_risk_score", "work_life_balance_risk", 
    "satisfaction_composite_score", "manager_stability_score", 
    "career_growth_risk_indicator", "burnout_risk_index"
]
corr_matrix = df[cols].corr()
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f", vmin=-1, vmax=1)
plt.title("Correlation Matrix of Custom Engineered Features")
plt.show()
```

## 3. Analyze Key Engineered Features

### 3.1 Overtime Risk Score
Let's see the distribution of `overtime_risk_score` for active vs departed employees.

```python
sns.boxplot(data=df, x="attrition", y="overtime_risk_score", palette="Set2")
plt.title("Overtime Risk Score by Attrition Status (0 = Active, 1 = Departed)")
plt.show()
```

### 3.2 Burnout Risk Index
Let's see if the composite `burnout_risk_index` is significantly higher for employees who left.

```python
sns.kdeplot(data=df, x="burnout_risk_index", hue="attrition", fill=True, common_norm=False, alpha=0.5)
plt.title("Burnout Risk Index Distribution by Attrition Status")
plt.show()
```
