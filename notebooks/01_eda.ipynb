# HR Attrition Intelligence - 01. Exploratory Data Analysis (EDA)

This notebook documents the exploratory analysis of the IBM HR Employee Attrition dataset. It explores employee demographics, satisfaction levels, compensation, and their relationship with voluntary attrition.

### Objectives:
1. Inspect data distribution and column types.
2. Analyze demographic and role-based factors affecting attrition.
3. Investigate the impact of compensation, overtime, and business travel.
4. Assess correlations between attributes.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set plotting style
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["font.size"] = 10
```

## 1. Load Data

```python
df_raw = pd.read_csv("../data/raw/employee_attrition_raw.csv")
print(f"Raw dataset shape: {df_raw.shape}")
df_raw.head()
```

## 2. Overall Attrition Rate

Let's see what percentage of employees have historically left the company.

```python
attrition_counts = df_raw["Attrition"].value_counts()
attrition_rate = df_raw["Attrition"].value_counts(normalize=True) * 100
print("Attrition Counts:")
print(attrition_counts)
print("\nAttrition Proportions (%):")
print(attrition_rate.round(2))
```

## 3. Demographics Analysis

Let's look at `Age` and `Gender` distributions relative to attrition.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

sns.kdeplot(data=df_raw, x="Age", hue="Attrition", fill=True, common_norm=False, alpha=0.5, ax=axes[0])
axes[0].set_title("Age Distribution by Attrition Status")

sns.countplot(data=df_raw, x="Gender", hue="Attrition", ax=axes[1])
axes[1].set_title("Attrition Count by Gender")

plt.tight_layout()
plt.show()
```

## 4. Work Environment & Role Volatility

Let's see which departments and roles have high turnover rates.

```python
plt.figure(figsize=(12, 6))
sns.countplot(data=df_raw, y="JobRole", hue="Attrition", order=df_raw["JobRole"].value_counts().index)
plt.title("Job Role Attrition Distribution")
plt.xlabel("Count")
plt.ylabel("Job Role")
plt.tight_layout()
plt.show()
```

## 5. Overtime & Travel Impacts

Let's evaluate how lifestyle/stress factors correlate with attrition.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

sns.barplot(data=df_raw.groupby("OverTime")["Attrition"].apply(lambda x: (x == "Yes").mean() * 100).reset_index(),
            x="OverTime", y="Attrition", ax=axes[0])
axes[0].set_title("Attrition Rate (%) by Overtime Status")
axes[0].set_ylabel("Attrition Rate (%)")

sns.barplot(data=df_raw.groupby("BusinessTravel")["Attrition"].apply(lambda x: (x == "Yes").mean() * 100).reset_index(),
            x="BusinessTravel", y="Attrition", order=["Non-Travel", "Travel_Rarely", "Travel_Frequently"], ax=axes[1])
axes[1].set_title("Attrition Rate (%) by Business Travel Frequency")
axes[1].set_ylabel("Attrition Rate (%)")

plt.tight_layout()
plt.show()
```
